```python
import asyncio
from streaming_ordering_chatbot.api.flows.conversation_flows_SK import (
    PreambleFlowSK,
    SummaryFlowSK,
    OrderAssistantFlowSK
)
from streaming_ordering_chatbot.api.models import Message

# Azure OpenAI configuration (using existing values)
ENDPOINT = "https://t-toluale-8955-resource.openai.azure.com/"
API_KEY = "AwkpoSc6XgvCOPrLv6FmBceTbaH2f5LlcgMHaXh0rB97bkQJJjRxJQQJ99BFACHYHv6XJ3w3AAAAACOG1Gzr"
DEPLOYMENT_NAME = "gpt-4o"

# Sample chat history
chat_history = [
    Message(role="system", content="Welcome! How can I help you today?"),
    Message(role="user", content="Hi! I'd like to know about your burgers."),
    Message(role="assistant", content="We have several delicious burger options available. Would you like to hear about our signature burgers?"),
    Message(role="user", content="Yes, please tell me about your cheeseburgers.")
]

# Sample current order
current_order = {
    "items": [
        {
            "code": "CZHB14SSN",
            "toppings": [
                {"code": "L", "amount": "1"},
                {"code": "T", "amount": "1"}
            ],
            "quantity": 1,
            "description": "1 - cheeseburger, quarter lb, single, sesame, normal with lettuce, tomato"
        }
    ]
}
```

```python
async def test_preamble_flow():
    """Test the preamble flow with different personalities"""
    # Initialize the flow
    preamble = PreambleFlowSK(ENDPOINT, API_KEY, DEPLOYMENT_NAME)
    
    # Test with default personality
    print("Testing Preamble Flow (Default Personality):")
    async for response in preamble(chat_history=chat_history[:1]):  # Just use the first message
        print(response, end='')
    print("\n")
    
    # Test with casual personality
    print("Testing Preamble Flow (Casual Personality):")
    async for response in preamble(
        chat_history=chat_history[:1],
        personality="Keep it chill and real, like you're talking to a buddy. Simple, street vibes—no fancy talk."
    ):
        print(response, end='')
    print("\n")

# Run the test
await test_preamble_flow()
```

```python
async def test_summary_flow():
    """Test the summary flow"""
    # Initialize the flow
    summary = SummaryFlowSK(ENDPOINT, API_KEY, DEPLOYMENT_NAME)
    
    print("Testing Summary Flow:")
    async for response in summary(chat_history=chat_history):
        print(response, end='')
    print("\n")

# Run the test
await test_summary_flow()
```

```python
async def test_order_assistant_flow():
    """Test the order assistant flow with different configurations"""
    # Initialize the flow
    assistant = OrderAssistantFlowSK(ENDPOINT, API_KEY, DEPLOYMENT_NAME)
    
    # Test with default personality
    print("Testing Order Assistant Flow (Default):")
    async for response in assistant(
        chat_history=chat_history,
        current_order=current_order
    ):
        print(response, end='')
    print("\n")
    
    # Test with casual personality
    print("Testing Order Assistant Flow (Casual Personality):")
    async for response in assistant(
        chat_history=chat_history,
        current_order=current_order,
        personality="Keep it chill and real, like you're talking to a buddy. Simple, street vibes—no fancy talk."
    ):
        print(response, end='')
    print("\n")
    
    # Test with brand personality
    print("Testing Order Assistant Flow (Brand Personality - Friendly Diner):")
    async for response in assistant(
        chat_history=chat_history,
        current_order=current_order,
        brand="friendly_diner"
    ):
        print(response, end='')
    print("\n")

# Run the test
await test_order_assistant_flow()
```
